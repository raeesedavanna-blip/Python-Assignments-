**Numpy Array Operations:**

```python
import numpy as np
```

```python
# 1creating 1D numpy array
temperature_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print(temperature_w1)
```

```python
# 2. Inspection and Properties:
print("Shape of the temperature_w1: ",temperature_w1.shape)
print("Data type of the temperature_w1: ",temperature_w1.dtype)
print("Number of Elements of the temperature_w1: ",temperature_w1.size)
```

```python
# 3. Array Operations:
temp_w1_f = (temperature_w1 * 9/5) + 32
print("Week 1 tmparature in Fahrenheit",temp_w1_f)
print("Max Temparature: ",np.max(temp_w1_f))
print("Min Temparature: ",np.min(temp_w1_f))
print("Mean Temparature: ",np.mean(temp_w1_f))




```

```python
# 4. Array Slicing and Indexing

print("First 3 days:", temperature_w1[:3])
print("Weekend (last 2 days):", temperature_w1[-2:])
print("Middle 3 days (days 3 to 5):", temperature_w1[2:5])
```

```python
# 5. Create a 2D Array
temperatures =np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])

print("2D array : ",temperatures)
```

```python
#6. Inspect and Slice the 2D Array:

print("Shape:", temperatures.shape)
print("Data Type:", temperatures.dtype)
print("Total Elements:", temperatures.size)
```

```python
#Week and Week End wise temperature
print("week 1 temparature :", temperatures[0, :])
print("week 1 > weekend temparature :",temperatures[0,-2:])
print("week 2 temparature :", temperatures[1, : ])
print("week 2 > weekend temparature :",temperatures[1,-2:])
```

**Pandas Series:**

```python
# 1. Creating Pandas Series:
import pandas as pd

marks_data = [95, 92, 89, 85, 80]
indices = ['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
marks = pd.Series(data=marks_data, index=indices)
print("Original Marks Series: ", marks)
```

```python
# 2. Indexing and Slicing:
print("Mark of the first rank student: ",marks.iloc[0])
print("Mark of the top 3 Ranks: ",marks.loc[['Rank1','Rank2','Rank3']])
print("Mark of the 3rd Rank ",marks.iloc[2])

mask = marks > 90
print("Ranks with Marks > 90: ",marks[mask])
```

```python
# 3. Manipulating Series:

#serirs can be updated like dictories so,
marks["Rank1"] = 100
print("\n Updated Marks Series: ")
print(marks)


#an entry can be removed  using .drop()
#marks = marks.drop('Rank5')
print("\n Rank5  removed data")
print("\n",marks)
```

```python
# CGPA
cgpa = marks/10
print("CGPA:\n",cgpa)
```

**Pandas DataFrame:**

```python
# 1. Creating Pandas DataFrame
import pandas as pd

data = {
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing',
                        'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
    'Region': ['North', 'South', 'North', 'East', 'West',
               'North', 'East', 'West', 'South', 'North'],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
}
transactions = pd.DataFrame(data)
print("Transactions DataFrame:\n", transactions)
```

```python
# 2. Data Exploration:
print("Transaction Head \n",transactions.head())
print("Transaction Tail \n",transactions.tail())
print("Transaction Shape\n",transactions.shape)
print("Transaction Columns\n",transactions.columns)
print("Transaction Data Type\n",transactions.dtypes)
```

```python
#Product Category and Amount Column
transactions[['ProductCategory','Amount']]
```

```python
#last 3 columns of the dataframe
transactions.iloc[:,-3:]
```

```python
#filter region Nort and Amount >200
filtered_transactions = transactions[(transactions['Region'] == 'North') & (transactions['Amount'] > 200)]
print(filtered_transactions)
```

```python
#value count for product category
transactions['ProductCategory'].value_counts()
```

```python
#Unique values in Region colums
transactions['Region'].unique()
```

```python
#Mean amount by Region
transactions.groupby('Region')['Amount'].mean()
```

```python
# 3. Manipulating the DataFrame:

#Amount modifaction for transcation id 102
transactions.loc[transactions['TransactionID']==102,'Amount'] =165
transactions
```

```python
#Creating new colum Discount by caclualting 10% of amount
transactions['Discount'] = transactions['Amount'] *.10
transactions
```

```python
#Removing row with trasaction id 109
transactions = transactions.drop(transactions[transactions['TransactionID']== 109] .index)
transactions
```

```python
#deleting the discount column from the DataFrame
transactions= transactions.drop(columns =['Discount'])
transactions
```
